```python
import pandas as pd
import numpy as np
import plotly.express as px #Creates charts with very little code.
import plotly.graph_objects as go #Gives full control over chart customization.
import plotly.io as pio #Handles input/output operations for Plotly figures.
import plotly.colors as colors #Provides predefined color palettes and color utilities.
pio.templates.default = "plotly_white"
```

```python
data = pd.read_csv("/content/drive/MyDrive/Colab Resource/Sample - Superstore.csv", encoding = 'latin-1') #Encoding helps to read special characters
```

```python
data.head()
```

```python
data['Order Date'].head()
```

```python
data.describe()
```

Converting Date-Time Column

```python
data['Order Date'] = pd.to_datetime(data['Order Date'])
data['Ship Date'] = pd.to_datetime(data['Ship Date'])
```

```python
data.info()
```

```python
data['Order Month'] = data['Order Date'].dt.month
data['Order Year'] = data['Order Date'].dt.year
data['Order Day of Week'] = data['Order Date'].dt.dayofweek
```

```python
data['Order Date'].head(10)
```

```python
data.head(10)
```

### 1. **Monthly Sales Analysis**

```python
sales_by_month = data.groupby('Order Month')['Sales'].sum().reset_index() #sales_by_month is a Pandas Series, not a DataFrame sales_by_month into a DataFrame using reset_index() so that 'Order Month' becomes an actual column
sales_by_month
```

```python
fig = px.line(sales_by_month ,
              x='Order Month',
              y='Sales',
              title= 'Monthly Sales Analysis')
fig.show()
```

## 2. **Sales Analysis by Category**

```python
sales_by_category = data.groupby('Category')['Sales'].sum().reset_index()
sales_by_category
```

```python
fig = px.pie(sales_by_category ,
              names='Category',
              values='Sales',
              hole = 0.4 ,
              title= 'Category Wise Sales Analysis')
fig.update_traces(textposition='inside', textinfo = 'percent + label')
fig.show()
```

### 3. **Sub-Category Wise Sales Analysis**

```python
sales_by_sub_category = data.groupby('Sub-Category')['Sales'].sum().reset_index()
sales_by_sub_category
```

```python
fig = px.bar(sales_by_sub_category ,
              x='Sub-Category',
              y='Sales',
              title= 'Sub-Category Wise Sales Analysis')
fig.show()
```

## 4. **Monthly Profit Analysis**

```python
monthly_profit = data.groupby('Order Month')['Profit'].sum().reset_index()
monthly_profit
```

```python
fig = px.line(monthly_profit ,
              x='Order Month',
              y='Profit',
              title= 'Monthly Profit Analysis')
fig.show()
```

```python
fig = px.bar(monthly_profit ,
              x='Order Month',
              y='Profit',
              title= 'Monthly Profit Analysis')
fig.show()
```
